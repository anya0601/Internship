````python
# Setting up the Target URL
from urllib import response


url = r"https://en.wikipedia.org/wiki/States_and_union_territories_of_India"

# Sending a GET request to the URL
response = requests.get(url).text # The `text` attribute of the response object contains the HTML content of the page

print(response) # Printing the HTML content of the page
````

```python
# 1. Check if a string contains only digits
def is_all_digits(s):
    return s.isdigit() # Built-in method returns True if all chars are numbers

# Call the function with a test string
```

```python
# 2. Find the median of a list of numbers
def find_median(numbers):
    nums = sorted(numbers)
    n = len(nums)
    mid = n // 2
    if n % 2 == 0:
        return (nums[mid - 1] + nums[mid]) / 2
    return nums[mid]

# Call the function with a test string
```

```python
# 3. Count words longer than N letters in a sentence
def count_long_words(sentence, n):
    words = sentence.split()
    # List comprehension: filter words where length > n, then get length of that list
    return len([word for word in words if len(word) > n])

# Call the function with a test string
```

```python
# 4. Swap two variables without using a temporary variable
def swap(a, b):
    a, b = b, a # Pythonic tuple unpacking
    return a, b

# Call the function with a test string
```

```python
# 5. Check if a given year is a leap year
def is_leap_year(year):
    # Divisible by 4, but if divisible by 100, must also be divisible by 400
    return (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)

# Call the function with a test string
```

```python
# 6. Find the common elements between two lists
def common_elements(list1, list2):
    return list(set(list1) & set(list2)) # Intersection of two sets


# Call the function with a test string
```

```python
# 7. Convert a list of (key, value) tuples into a dictionary
def tuples_to_dict(tuple_list):
    return dict(tuple_list)

# Call the function with a test string
```

```python
# 8. Compute the running/cumulative sum of a list
def cumulative_sum(numbers):
    total = 0
    res = []
    for n in numbers:
        total += n
        res.append(total)
    return res

# Call the function with a test string
```

```python
# 9. Check if a string has balanced parentheses
def is_balanced(s):
    stack = []
    for char in s:
        if char == '(': stack.append(char)
        elif char == ')':
            if not stack: return False
            stack.pop()
    return len(stack) == 0

# Call the function with a test string
```

```python
# 10. Find the most frequently occurring character in a string

import collections
def most_frequent(s):
    # Counter creates a frequency map, most_common(1) returns [(char, count)]
    return collections.Counter(s).most_common(1)[0][0]

# Call the function with a test string
```

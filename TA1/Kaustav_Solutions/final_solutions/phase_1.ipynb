```python
# 1. Check if a string contains only digits
def is_all_digits(s):
    return s.isdigit() # Built-in method returns True if all chars are numbers

# Call the function with a test string
print(is_all_digits("12345"))  # Returns True
print(is_all_digits("123a45")) # Returns False
print(is_all_digits(""))  # Returns False, as empty string has no digits
```

```python
# 2. Find the median of a list of numbers
def find_median(numbers):
    nums = sorted(numbers)
    n = len(nums)
    mid = n // 2
    if n % 2 == 0:
        return (nums[mid - 1] + nums[mid]) / 2
    return nums[mid]

# Call the function with a test string

print(find_median([3, 1, 2]))  # Returns 2
print(find_median([3, 1, 2, 4])) # Returns 2.5
print(find_median([1, 9, 2.5 , 7, 3, 4])) # Returns 3.5
```

```python
# 3. Count words longer than N letters in a sentence
def count_long_words(sentence, n):
    words = sentence.split()
    # List comprehension: filter words where length > n, then get length of that list
    return len([word for word in words if len(word) > n])

# Call the function with a test string
print(count_long_words("A quick brown fox jumps over the lazy dog", 4))  # Returns 2 ("quick", "jumps")
print(count_long_words("Sovereign Socialist Secular Democratic Republic", 5))  # Returns 2 ("Sovereign", "Democratic")
```

```python
# 4. Swap two variables without using a temporary variable
def swap(a, b):
    a, b = b, a # Pythonic tuple unpacking
    return a, b

# Call the function with a test string

print(swap(5, 10))  # Returns (10, 5)
print(swap("hello", "world"))  # Returns ("world", "hello")
print(swap([1, 2], [3, 4]))  # Returns ([3, 4], [1, 2])
```

```python
# 5. Check if a given year is a leap year
def is_leap_year(year):
    # Divisible by 4, but if divisible by 100, must also be divisible by 400
    return (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)

# Call the function with a test string

print(is_leap_year(2020))  # Returns True
print(is_leap_year(1900))  # Returns False
print(is_leap_year(2000))  # Returns True
print(is_leap_year(2021))  # Returns False
```

```python
# 6. Find the common elements between two lists
def common_elements(list1, list2):
    return list(set(list1) & set(list2)) # Intersection of two sets


# Call the function with a test string

print(common_elements([1, 2, 3, 4], [3, 4, 5, 6]))  # Returns [3, 4]
print(common_elements(['apple', 'banana', 'cherry'], ['banana', 'kiwi', 'apple']))  # Returns ['banana', 'apple']
print(common_elements([1, 2, 3], [4, 5, 6]))  # Returns []
```

```python
# 7. Convert a list of (key, value) tuples into a dictionary
def tuples_to_dict(tuple_list):
    return dict(tuple_list)

# Call the function with a test string
print(tuples_to_dict([('Id', 12452), ('Name', 'Ananya'), ('Department', "Economics")]))  # Returns {'Id': 12452, 'Name': 'Ananya', 'Department': 'Economics'}
```

```python
# 8. Compute the running/cumulative sum of a list
def cumulative_sum(numbers):
    total = 0
    res = []
    for n in numbers:
        total += n
        res.append(total)
    return res

# Call the function with a test string
print(cumulative_sum([1, 2, 3, 4, 5, 6])) # Returns [1, 3, 6, 10, 15, 21]
print(cumulative_sum([10, 20, 30])) # Returns [10, 30, 60]
print(cumulative_sum([-1, 1, -1, 1])) # Returns [-1, 0, -1, 0]
```

```python
# 9. Check if a string has balanced parentheses
def is_balanced(s):
    stack = []
    data = {'(': ')', '{': '}', '[': ']'}
    for char in s:
        if char in data:
            stack.append(char)
        elif char in data.values():
            if not stack or data[stack.pop()] != char:
                return False
    return len(stack) == 0

# Call the function with a test string
print(is_balanced("()"))  # Returns True
print(is_balanced("({[})"))  # Returns False
print(is_balanced(")("))  # Returns False
print(is_balanced("[]{()}()")) # Returns True
```

```python
# 10. Find the most frequently occurring character in a string

import collections
def most_frequent(s):
    # Counter creates a frequency map, most_common(1) returns [(char, count)]
    return collections.Counter(s).most_common(1)[0][0]

# Call the function with a test string
print(most_frequent("hello world"))  # Returns 'l'
print(most_frequent("aabbbcc"))      # Returns 'b'
print(most_frequent("abcde"))        # Returns 'a' (or any character, as all have same frequency)
```
